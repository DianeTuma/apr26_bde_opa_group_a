Before starting with feature Engineering and create new Columns as Indicators for a better Prediction. we have first to look after the Target feature which we don't have until now. The Goal of our ML model will be to predict if we can sell or not. So a classification Problem. Let now beginn with.

### Creation of the Target Feature ####

```python

LOOKAHEAD = 3 # we look at 3 klines in the future
df['future_close'] = df['price'].shift(-LOOKAHEAD)
df['target'] = (df['future_close'] > df['price']).astype(int)
df = df.drop(columns=['future_close'])

# On supprime les 'LOOKAHEAD' dernières lignes du dataset car elles n'ont pas de futur
df = df.dropna(subset=['target']).reset_index(drop=True)
df.head(3)
# target : 1 -> buy / 0 -> sell
```

We Seplit now 

```python
# 2. SPLIT TRAIN / TEST
split_index = int(len(df) * 0.80)
df_train = df.iloc[:split_index].copy().reset_index(drop=True)
df_test = df.iloc[split_index:].copy().reset_index(drop=True)
```

### Feature Engineering ###

```python
import pandas_ta as ta

def add_indicators(data):
    # Create a copy to prevent SettingWithCopyWarning from Pandas
    df_feat = data.copy()
    
    # --- 1. TRENDS ---
    # Exponential Moving Averages (20/50 crossover strategy)
    df_feat['ema_20'] = ta.ema(df_feat['price'], length=20)
    df_feat['ema_50'] = ta.ema(df_feat['price'], length=50)
    # Moving Average Convergence Divergence (Trend + Momentum)
    macd = ta.macd(df_feat['price'], fast=12, slow=26, signal=9)
    df_feat = pd.concat([df_feat, macd], axis=1)
    
    # --- 2. VOLATILITY ---
    # Bollinger Bands (BB)
    bbands = ta.bbands(df_feat['price'], length=20, std=2)
    df_feat = pd.concat([df_feat, bbands], axis=1)
    # Average True Range (ATR) - Excellent for determining stop-loss size
    df_feat['atr'] = ta.atr(df_feat['high'], df_feat['low'], df_feat['price'], length=14)
    
    # --- 3. MOMENTUM ---
    # Relative Strength Index (RSI)
    df_feat['rsi'] = ta.rsi(df_feat['price'], length=14)
    
    # --- 4. SHORT-TERM MARKET DYNAMICS ---
    # Period-to-period price return
    df_feat['returns'] = df_feat['price'].pct_change()
    # Short-term volume momentum (current volume compared to its 5-period rolling mean)
    df_feat['vol_momentum'] = df_feat['volume'] / df_feat['volume'].rolling(window=5).mean()
    
    # --- CLEANING ---
    # Lagged indicators (like EMA 50) generate NaN values in the initial rows.
    # We drop them to prevent Machine Learning models from crashing.
    df_feat = df_feat.dropna().reset_index(drop=True)
    
    return df_feat
```

```python
df_train_features = add_indicators(df_train)
df_test_features = add_indicators(df_test)
df_train_features.head(3)
```

Let now transform the data so that the Model will not consider a Attribut more important than another

```python
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score
```

Let separate X and Y for training

```python
columns_to_drop = ['timestamp_open', 'timestamp_close', 'market', 'target']

X_train = df_train_features.drop(columns=columns_to_drop)
y_train = df_train_features['target']

X_test = df_test_features.drop(columns=columns_to_drop)
y_test = df_test_features['target']
```

Scaling of data

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

### Model Fitting and Probability

```python
# Initialize and train the model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train_scaled, y_train)

# Get standard binary predictions (0 or 1)
y_pred = model.predict(X_test_scaled)

# Get precise probabilities of buying (class 1)
buy_probabilities = model.predict_proba(X_test_scaled)[:, 1]
```

### Model Evaluation

```python
# Check the evaluation
print("Accuracy Score:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

### Results

```python
# We build a clean DataFrame to visualize the results
results_df = pd.DataFrame({
    'Actual_Target': y_test,
    'Model_Prediction': y_pred,
    'Buy_Probability': buy_probabilities
})

# we define our confidence threshold
CONFIDENCE_THRESHOLD = 0.65

# Create a custom signal: 1 if probability > threshold, else 0
results_df['Filtered_Signal'] = (results_df['Buy_Probability'] > CONFIDENCE_THRESHOLD).astype(int)

# Let's see how many trades we actually take with this filter
print(f"Total signals without filter: {results_df['Model_Prediction'].sum()}")
print(f"Total signals with 65% filter: {results_df['Filtered_Signal'].sum()}")
```
